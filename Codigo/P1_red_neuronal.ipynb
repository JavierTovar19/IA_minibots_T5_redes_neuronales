```python
import numpy as np

# Función sigmoide
def sigma(z):
    return 1/(1+np.exp(-z))

# Derivada de la sigmoide
def sigma_deriv(z):
    s = sigma(z)
    return s*(1-s)

# Datos de entrada
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])

# 👉 Cambiar aquí según compuerta:
# NAND
Y = np.array([[1],[1],[1],[0]])

# XOR
# Y = np.array([[0],[1],[1],[0]])

np.random.seed(0)

# Inicialización de pesos y sesgos
W1 = np.random.rand(2,4)
b1 = np.random.rand(1,4)

W2 = np.random.rand(4,4)
b2 = np.random.rand(1,4)

W3 = np.random.rand(4,1)
b3 = np.random.rand(1,1)

eta = 0.5
epochs = 10000

for i in range(epochs):

    # Forward propagation
    Z1 = np.dot(X,W1) + b1
    A1 = sigma(Z1)

    Z2 = np.dot(A1,W2) + b2
    A2 = sigma(Z2)

    Z3 = np.dot(A2,W3) + b3
    Y_hat = sigma(Z3)

    # Error
    error = Y - Y_hat

    # Backpropagation
    dZ3 = error * sigma_deriv(Z3)
    dW3 = np.dot(A2.T,dZ3)
    db3 = np.sum(dZ3,axis=0)

    dZ2 = np.dot(dZ3,W3.T) * sigma_deriv(Z2)
    dW2 = np.dot(A1.T,dZ2)
    db2 = np.sum(dZ2,axis=0)

    dZ1 = np.dot(dZ2,W2.T) * sigma_deriv(Z1)
    dW1 = np.dot(X.T,dZ1)
    db1 = np.sum(dZ1,axis=0)

    # Actualización de pesos
    W3 += eta*dW3
    b3 += eta*db3

    W2 += eta*dW2
    b2 += eta*db2

    W1 += eta*dW1
    b1 += eta*db1

print("Resultado final:")
print(np.round(Y_hat,3))
```
